```python

# Import required libraries

import tensorflow as tf

from tensorflow.keras.datasets import cifar10
from tensorflow.keras.applications import VGG16
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Flatten, Dense
from tensorflow.keras.utils import to_categorical

print("All libraries imported successfully!")
```

```python
# Step 1: Load CIFAR-10 dataset
(itrain, ltrain), (itest, ltest) = cifar10.load_data()
```

```python
# Step 2: Load pre-trained VGG16 model

basem = VGG16(
    weights='imagenet',
    include_top=False,
    input_shape=(32, 32, 3)
)

print("VGG16 model loaded successfully!")
```

```python
# Step 4: Create new model on top of VGG16
semodel = Sequential()
semodel.add(basem)
semodel.add(Flatten())
semodel.add(Dense(256, activation='relu'))
semodel.add(Dense(10, activation='softmax'))  # CIFAR-10 → 10 classes
```

```python
# Step 5: Compile the model
semodel.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
# Step 6: Train the model
history = semodel.fit(
    itrain, ltrain,
    epochs=10,
    batch_size=32,
    validation_data=(itest, ltest)
)
```

```python
# Step 7: Evaluate model
loss, acc = semodel.evaluate(itest, ltest)

print("\nTest accuracy:", acc)
```
